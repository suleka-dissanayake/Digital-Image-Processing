```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
image_path = "sample_color.jpg"
img = np.array(Image.open(image_path))
```

```python
plt.figure(figsize=(6,5))

if img.ndim == 2:
    plt.imshow(img, cmap='gray')
else:
    plt.imshow(img)
    
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
height = img.shape[0]
width = img.shape[1]
dimension = img.shape[2]

print("Height:",height)
print("Width:",width)
print("Dimension:",dimension)
```

```python
if img.ndim == 3:
    print("Color image")
else:
    print("Gray image")
```

```python
red_channel = img[:,:,0]
green_channel = img[:,:,1]
blue_channel = img[:,:,2]
```

```python
plt.figure(figsize=(6,5))
plt.subplot(1,3,1)
plt.imshow(red_channel, cmap='gray')
plt.title('Red Channel')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(green_channel, cmap='gray')
plt.title('Green Channel')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(blue_channel, cmap='gray')
plt.title('Blue Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
row = 50
col = 50
print("Pixel before change:", img[row, col])
```

```python
modified_img = img.copy()
modified_img[row, col] = [255, 0, 0]
print("Pixel after change:", modified_img[row, col])
```

```python
roi_img = img.copy()
roi_img[50:150, 50:150] = 0
```

```python
plt.figure(figsize=(6,5))
plt.imshow(roi_img)
plt.title("Image with black square ROI")
plt.axis('off')
plt.show()
```
