```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("sample_image.jpg", cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(3,4))
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
# brightness
```

```python
brighter = img + 60
brighter = np.clip(brighter, 0, 255)
brighter = brighter.astype(np.uint8)

plt.figure(figsize=(3,4))
plt.imshow(brighter, cmap='gray')
plt.title('Brightened Image')
plt.axis('off')
plt.show()
```

```python
# gamma

normalized = img / 255.0

gamma_05 = np.power(normalized, 0.5) * 255
gamma_05 = np.clip(gamma_05, 0, 255)
gamma_05 = gamma_05.astype(np.uint8)

gamma_10 = np.power(normalized, 1.0) * 255
gamma_10 = np.clip(gamma_10, 0, 255)
gamma_10 = gamma_10.astype(np.uint8)

gamma_20 = np.power(normalized, 2.0) * 255
gamma_20 = np.clip(gamma_20, 0, 255)
gamma_20 = gamma_20.astype(np.uint8)

plt.figure(figsize=(15,4))
plt.subplot(1,3,1)
plt.imshow(gamma_05, cmap='gray')
plt.title("Gamma 0.5 (manual)")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(gamma_10, cmap='gray')
plt.title("Gamma 1.0 (manual)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(gamma_20, cmap='gray')
plt.title("Gamma 2.0 (manual)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# log transformation
```

```python
img_float = img.astype(np.float32)
img_float = np.clip(img_float, 0, 255)

c = 255.0 / np.log(1+255)
log_transformed = c * np.log1p(img_float)

log_transformed = c * np.log1p(img_float)
log_transformed = np.clip(log_transformed, 0, 255)
log_transformed = log_transformed.astype(np.uint8)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img.astype(np.uint8), cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(log_transformed, cmap='gray')
plt.title('Log Transformed')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
color_img = cv2.imread("images.jpg")
colorimg = cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB)
plt.imshow(colorimg)
plt.axis('off')
plt.show()
```

```python
red_channel = colorimg[:,:,0]
green_channel = colorimg[:,:,1]
blue_channel = colorimg[:,:,2]

plt.figure(figsize=(6,5))
plt.subplot(1,3,1)
plt.imshow(red_channel, cmap='gray')
plt.title('Red Channel')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(green_channel, cmap='gray')
plt.title('Green Channel')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(blue_channel, cmap='gray')
plt.title('Blue Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```
